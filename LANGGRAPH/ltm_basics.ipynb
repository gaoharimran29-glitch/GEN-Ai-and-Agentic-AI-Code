In this notebook we will learn about how a memory store works for long term memory in lanngraph. We will create a memorystore , then performal retrieval and semantic search. This helps you get familiar with basic codes memory stores then in next we will implement it in real chatbot

```python
from langgraph.store.memory import InMemoryStore
## This is a memory storage system provided by LangGraph. It stores memories: temporarily in RAM
```

```python
store = InMemoryStore()
```

```python
namespace1 = ("user" , "u1")
## It is just like creating a folder in memory as shown below 

## STORE
## │
## ├── ("user", "u1")
## │
## └── ("user", "u2")
```

Creating Memories

```python
# Adding memories

store.put(namespace1, "1", {"data": "User likes pizza"})
## It reuqires three things (folder name (namespace) , key , data)
## This key is assigned with data for reference

store.put(namespace1, "2", {"data": "User prefers dark mode"})
```

```python
# another namespace
namespace2 = ("user", "u2")

# Add memories
store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

Retrieving Memories

```python
# store.get(namespace, key)
store.get(namespace2, "2")
```

Retrieving all memories

```python
items = store.search(namespace2)

for item in items:
    print(item.value)
```

Semantic Search :- So in production grade system obviously LLM will not know the key id to fetch the relevant user info from LTM . Also retrieving all memory and giving it to LLM is not good because then LLM will get confuse and it also increase the context window. So we use semantic search to retrieve the relevant info from the LTM

```python
from langchain_huggingface.embeddings import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2") 
## first of all we need an embedding model to perform semantic search after making embeddings
```

```python
store = InMemoryStore(index={'embed': embedding, 'dims':1536}) ## We have to initialize memory with emebedding model like this
```

```python
namespace = ('users', 'u1') ## this again creates the folder inside the memory
```

```python
store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})

## Now we will perform semantic search retrieval over these memory data
```

```python
items = store.search(namespace, query="what is the user currently learning", limit=1)
## limit parameter sets the number of outputs. if limit is 1 we will get only result of retrieval

for item in items:
    print(item.value)
```

```python
items = store.search(namespace, query="what are user's preferences", limit=3)

for item in items:
    print(item.value)
```
